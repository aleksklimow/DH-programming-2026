Некоторые определения
Если кратко, то коллокации - это устойчивые выражения, состоящие из двух и более слов. Устойчивые - значит, что они часто используются вместе; ***"the company words keep"***, по неформальному определению Джона Фёрта (1957). Коллокации позволяют нам смотреть на слова не изолированно, а в контексте, без чего невозможно заниматься как чисто лингвистическим, так и антропологическим, историческом и вообще гуманитарным в широком смысле исследованием.

**Коллокации** - это последовательности слов, которые встречаются вместе чаще, чем можно ожидать случайно, в естественном языке. Они представляют собой "привычные" или "устойчивые" комбинации, где значение целого часто не сводится к сумме частей (например, "strong tea" vs. "powerful tea" - первое естественно, второе звучит странно). В отличие от идиом (где значение полностью непрозрачно, как "бить баклуши"), коллокации обладают определённой степенью **композициональности**: их можно понять из компонентов, но они фиксированы контекстом (= значение больших единиц равно сумме значений меньших единиц). Единицы, составляющие коллокации называются **коллокатами**.

Проанализируйте две фразы:
1. Железная кровать
2. Железная дорога
3. Съел собаку...
   
И на английском:
1. Johanna always kicks the bucket. Stop her! I hate the sound.
2. Poor Johnny. He finally kicked the bucket...

Есть ли у вас примеры композициональных и некомпозициональных выражений?

Некомпозициональные выражения **идиоматичны**: их значение не выводится из суммы значений частных, они устойчивы и воспроизводятся целиком, они естественны и зачастую экспрессивны, обусловлены культурой.

Несколько цитат из книги Майкла Стаббса **"Words and Phrases"** (2002):
1. Our **knowledge of a language is not only a knowledge of individual words, but of their predictable combinations**, and of the cultural knowledge which these combinations often encapsulate.
2. Most everyday uses of language are much more flexible, although we are rarely, if ever, entirely free in what it is appropriate to say. Words occur in expected sequences in phrases and texts.
3. In isolation, many individual words are ambiguous or indeterminate in meaning, but this hardly ever troubles us in practice, because the phrases in which they occur are not ambiguous.
4. In other words, cases of apparent multiple ambiguity at word level are usually illusory: they dissolve in context. Combinations of words in phrases are therefore a good candidate for the basic semantic unit of language in use. The word surgery conveys a rather general meaning: 'something to do with medicine'. It is the phrase which conveys the precise meaning.
5. It is impossible to observe the meaning of the word: **meaning is an invisible (arguably mental) phenomenon**. Howerver, it is quite possible to observe evidence from which meanings can be reliably inferred. A major type of evidence of the meaning of a word is the other words round about it, especially repeated patterns of co-occurence.
6. There is some consensus, but no total agreement, that **significant collocates are usually found within a span of 4:4** (Jones and Sinclair 1974).

Коллокации помогают анализировать стиль авторов, эволюцию языка, культурные паттерны и семантику в больших корпусах. Например, уже сталкивались на курсе по анализу текста с использованием коллокаций как основы для построения эмбеддингов. Пара примеров:

Например, в этой (https://academic.oup.com/dsh/article/36/3/736/6034823?) статье извлекаются данные о растениях Озёрного края в Англии, их синонимах, местоположениях и изменениях во времени, чтобы создать структурированный датасет для геотемпорального анализа в GIS. Анализ коллокаций здесь выступает как инструмент извлечения информации о местоположении конкретного растения: на определенном спэне в тексте выявлялись термины, частотно употребляющеся вместе, так были выявлены пары названий растений и локация для помещения на карту.

В статье "Дальнее чтение советских дневников" анилизируется 380 тысяч дневниковых записей, коллокационный анализ там имеет одно из ключевых значений для исследования общественного дискурса.

Ну и, конечно, коллокации имеют самое широкое распротранение в лингвистических и текстовых исследованиях - авторский стиль, авторские выражения, типичные выражения в разных типах речевой деятельности, жанрах и видах текстов, уровнях владения иностранным языком и многое другое.

Коллокации размера n можно считать самым простым образом - посредством подсчёта абсолютной или относительной частоты - относительно количества n-грамм в тексте. Однако помимо это существуют **меры ассоциации** (или коллокационные меры). Таких мер бесчисленное множество, дающих несколько разный результат из-за разного принципа подсчёта.

Пример из В. Брежины:

Если мы исследуем коллокаты слова love, то в этом случае love является **узлом**, т.е. искомым словом. **Коллокатами** в таком случае будут называться слова, которые со-встречаются с узлом в определённом **диапазоне** (коллокационном **окне**), в данном случае это 1 слово слева и одно слово справа. В этом примере с love со-встречаются:
1. my - 3,
2. is - 2,
3. in - 1,
4. am - 1,
5. will -1,
6. thee - 1,
7. only - 1,
8. and -1,
9. thou - 1. 

Однако можно ли делать на основании этого, что my cильно ассоциирован с love? Как это понять?
1. Можно отранжировать по частоте и опираться на ранг;
2. Можно сгенерировать с теми же словами случайный текст и сравнить с рангом в случайном порядке;
3. Можно использовать более сложные коллокационные метрики.

## Займёмся практической частью.

```python
import nltk
from nltk.collocations import BigramCollocationFinder, BigramAssocMeasures, TrigramCollocationFinder, TrigramAssocMeasures
from nltk.tokenize import word_tokenize
from nltk.tokenize import sent_tokenize
from nltk.corpus import stopwords
import pymorphy3 
import re
import string
from collections import Counter
import requests


nltk.download('punkt', quiet=True)
nltk.download('stopwords', quiet=True)

morph = pymorphy3.MorphAnalyzer()  
stop_words = set(stopwords.words('russian'))  
```

Зададим несколько функций, которые будут обрабатывать наши тексты для большей объективности и чистоты: очистим от стоп-слов и лемматизируем. 
Функция **normalize** лемматизирует наш текст иоставит от них только кириллические слова и числа, приведет текст к нижнему регистру, а затем выкинет все слова короче 3 букв.  
Также нам нужен **sent_tokenize** из nltk.
**ngrammer** cобирает нграммы (заданной величины) и их частоты.

```python
stops = set(stopwords.words('russian') + ["это", "весь"]) 

def normalize(text):
    normalized_text = [morph.parse(token)[0].normal_form for token in word_tokenize(text.lower()) if len(token) > 2 and token not in stops]  
    return normalized_text

def ngrammer(tokens, n=2):
    ngrams = []
    tokens_cl = [token for token in word_tokenize(tokens.lower()) if token not in stops]
    for i in range(0, len(tokens_cl)-n+1):
        ngrams.append(tuple(tokens_cl[i:i+n]))
    return ngrams
```

Нас скорее интересуют случаи, когда слова в большинстве случаев встречаются вместе. Для этого мы можем придумать какие-нибудь формулы, учитывающие частоты слов по отдельности и общую частоту.  

Самый простой способ - взять количество упоминаний биграммы и поделить на сумму количеств упоминаний слов по отдельности.  

Такая формула называется PMI.

```python
def scorer_simple(word_count_a, word_count_b, bigram_count, *args): #*args позволяет подать в функцию любое количество аргументов, так что в данном случае, если пользователь подаст лишнее, то лишнее будет проигнорировано 
    try:
        score = bigram_count / (word_count_a + word_count_b)
    
    except ZeroDivisionError:
        return 0
    
    return score

#Сделаем функцию, которая будет делать счетчики для слов и биграмм
def collect_stats(corpus, stops):
    ## соберем статистики для отдельных слов
    ## и биграммов
    
    unigrams = Counter()
    bigrams = Counter()
    
    for sent in corpus:
        unigrams.update(word_tokenize(sent))
        bigrams.update(ngrammer(sent))
    
    return unigrams, bigrams

#И функцию, которая пройдет по всем биграммам и вычислит для них нашу метрику.
def score_bigrams(unigrams, bigrams, scorer, threshold=0.2, min_count=2):
    ## посчитаем метрику для каждого нграмма
    bigram2score = Counter()
    len_vocab = len(unigrams)
    for bigram in bigrams:
        score = scorer(unigrams[bigram[0]], unigrams[bigram[1]], 
                       bigrams[bigram], len_vocab, min_count)
        
        ## если метрика выше порога, добавляем в словарь
        if score > threshold:
            bigram2score[bigram] = score
    
    return bigram2score
```

```python
unigrams, bigrams = collect_stats(corpus_sent_clean, stops)
```

```python
bigram2score = score_bigrams(unigrams, bigrams, scorer_simple)
len(bigram2score)
```

# Меры ассоциации, доступные в nltk

1. **chi_sq**: рассчитывается как хи-квадрат, её сила заключается в измерении значимости на основе отклонения от ожидаемых частот при предположении независимости с асимптотическим распределением χ², но её недостаток - чувствительность к размеру выборки, малым частотам в ячейках и искаженным выборкам.
2. **dice**: рассчитывается как коэффициент Дайса, её сила - нормализация совместной встречаемости для сбалансированного сравнения пересечения в разных контекстах, но её недостаток - игнорирование статистической независимости или значимости, что приводит к предпочтению частых терминов независимо от истинной силы коллокации.
3. **fisher**: рассчитывается с использованием точного теста Фишера для p-значения в таблице сопряженности, её сила - подходит для малых выборок или разреженных данных без асимптотических приближений.
4. **jaccard**: рассчитывается как индекс Жаккара, её сила - простая мера сходства множеств для бинарных данных о присутствии/отсутствии, но её недостаток - игнорирование частотности и статистической зависимости.
5. **likelihood_ratio**: рассчитывается как статистика теста отношения правдоподобия, её сила - меньшая чувствительность к малым или искаженным выборкам по сравнению с хи-квадрат для крупномасштабных вычислений.
6. **mi_like**: рассчитывается как вариант взаимной информации без логарифма, часто с экспонентой в числителе, её сила - захват отклонения от независимости гибким, вероятностным способом, похожим на MI, но её недостаток - предвзятость к низкочастотным событиям и отсутствие логарифмического масштабирования, что может переоценивать редкие пары.
7. **phi_sq**: рассчитывается как phi-квадрат, квадрат коэффициента корреляции Пирсона, её сила - предоставление ограниченного размера эффекта для ассоциации в таблицах сопряженности, но её недостаток - уязвимость к разреженным или несбалансированным данным.
8. **pmi**: рассчитывается как точечная взаимная информация, очень хорошо оценивает ассоциацию, но склонна переоценивать редкие, ошибочные коллокации.
9. **poisson_stirling**: рассчитывается как приближение лог-правдоподобия с использованием формулы Стирлинга для редких событий , её сила - хорошая обработка низкочастотных коллокаций, но её недостаток - зависимость от модели Пуассона, которая может не подходить ко всем распределениям данных, и ограниченная применимость к очень редким коллокациям.
10. **raw_freq**: рассчитывается просто как наблюдаемое количество совместных встречаемостей, её сила - интуитивность и простота вычисления, отражающая эмпирическую частотность, но её недостаток - сильное влияние общей частоты терминов без контроля за случайными совместными встречаемостями, что часто включает бессмысленные высокочастотные пары вроде стоп-слов.
11. **student_t**: рассчитывается как статистика t-критерия Стьюдента при гипотезе независимости, её сила - нормализация отклонения стандартной ошибкой для тестирования гипотезы независимости униграмм, но минус - предположение нормальности данных, при разреженных данных приводит к меньшей устойчивости по сравнению с хи-квадрат или PMI без фильтрации.
